```python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer

# Example corpus: replace this with your actual documents list or data loader
documents = [
    "Natural language processing and machine learning are exciting fields.",
    "TF-IDF is a numerical statistic that reflects how important a word is to a document.",
    "Machine learning models require clean data and robust feature engineering.",
    "Python is a popular programming language for data science and artificial intelligence."
]

# 1. Initialize the TF-IDF vectorizer (removes standard English stop words)
vectorizer = TfidfVectorizer(stop_words='english')

# 2. Fit and transform the documents into a TF-IDF matrix
tfidf_matrix = vectorizer.fit_transform(documents)

# 3. Retrieve the feature names (vocabulary words)
feature_names = vectorizer.get_feature_names_out()

# 4. Extract and print the top-5 terms for each document
top_n = 5
for i, doc in enumerate(documents):
    # Get the sparse/dense row for the current document
    row = tfidf_matrix.toarray()[i]

    # Sort indices by TF-IDF score in descending order
    sorted_indices = np.argsort(row)[::-1]

    # Filter out terms with a score of 0 and take the top N
    top_terms = [(feature_names[idx], row[idx]) for idx in sorted_indices[:top_n] if row[idx] > 0]

    print(f"Document {i + 1} Summary (Top Terms):")
    for term, score in top_terms:
        print(f"  - {term}: {score:.4f}")
    print("-" * 40)
```
