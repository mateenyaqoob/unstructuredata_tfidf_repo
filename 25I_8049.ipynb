```python
import sys, os
os.chdir("/content/tf-idfexamples/unstructuredata_tfidf_repo")
sys.path.insert(0, os.getcwd())

from src.load_data import load_documents
from sklearn.feature_extraction.text import TfidfVectorizer

# Step 1: load into notebook variables
titles, docs = load_documents()
print(f"Loaded {len(docs)} documents.")

# Step 2: build TF-IDF matrix
vectorizer = TfidfVectorizer(stop_words="english")
tfidf_matrix = vectorizer.fit_transform(docs)

print("Matrix shape:", tfidf_matrix.shape)
print("Vocabulary size:", len(vectorizer.get_feature_names_out()))
```

```python
import pandas as pd

# Convert sparse matrix to a labeled DataFrame
tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    index=titles,
    columns=vectorizer.get_feature_names_out()
)

```

```python
# Sample: first 3 documents x first 10 terms
print("Sample of TF-IDF matrix (3 docs x 10 terms):")
display(tfidf_df.iloc[:3, :10].round(3))

```

```python
# Top-5 terms per document
print("\nTop-5 TF-IDF terms per document:")
for title in tfidf_df.index:
    top5 = tfidf_df.loc[title].sort_values(ascending=False).head(5)
    terms = ", ".join(f"{term} ({score:.3f})" for term, score in top5.items())
    print(f"- {title}: {terms}")
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def rank_documents(query, vectorizer, tfidf_matrix, titles):
    """Transform the query with the SAME vectorizer and rank documents by cosine similarity."""
    query_vec = vectorizer.transform([query])            # transform only, do not fit again
    scores = cosine_similarity(query_vec, tfidf_matrix).flatten()
    ranking = sorted(zip(titles, scores), key=lambda x: x[1], reverse=True)
    return ranking

queries = [
    "renewable energy and climate policy",
    "password security and phishing",
    "healthy sleep exercise and diet",
]

for q in queries:
    ranking = rank_documents(q, vectorizer, tfidf_matrix, titles)
    print(f'\nQuery: "{q}"')
    for rank, (title, score) in enumerate(ranking[:3], start=1):
        print(f"   {rank}. {title:<22} {score:.4f}")
```
