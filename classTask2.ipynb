```python
import os
import sys
import nltk
from collections import Counter

# 1. NLTK required packages download karein
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

# 2. Agar repo clone nahi hui toh clone karein
if not os.path.exists('unstructuredata_tfidf_repo'):
    !git clone https://github.com/mateenyaqoob/unstructuredata_tfidf_repo.git

# 3. Path set kar ke documents import karein
sys.path.append('unstructuredata_tfidf_repo')
from data.documents import DOCUMENTS

# 4. NLTK ke English stopwords ka set lein
stop_words = set(stopwords.words('english'))

all_removed_stopwords = []

# 5. Har document ko tokenize kar ke sirf stopwords extract karein
for doc in DOCUMENTS:
    tokens = word_tokenize(doc.lower())
    # Sirf woh tokens rakhein jo alphabetic hon aur stopwords ki list mein mojood hon
    doc_stopwords = [word for word in tokens if word.isalpha() and word in stop_words]
    all_removed_stopwords.extend(doc_stopwords)

# 6. Counter use kar ke har stopword ki frequency nikal lein
stopword_counts = Counter(all_removed_stopwords)

print("=== ALL STOPWORDS & THEIR FREQUENCIES ===")
print(f"Total Stopwords Found Across All Documents: {len(all_removed_stopwords)}\n")

# Sab se zyada aane wale stopwords pehle print honge (most common order)
for word, freq in stopword_counts.most_common():
    print(f"'{word}': {freq} times")
```

```python
import os
import sys
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

# 1. Agar repo clone nahi hui toh clone karein
if not os.path.exists('unstructuredata_tfidf_repo'):
    !git clone https://github.com/mateenyaqoob/unstructuredata_tfidf_repo.git

# 2. Path set kar ke documents import karein
sys.path.append('unstructuredata_tfidf_repo')
from data.documents import DOCUMENTS

# 3. Document titles/names define karein (Index ke liye)
doc_titles = [
    "Doc 1: AI Healthcare",
    "Doc 2: Climate Change",
    "Doc 3: Basketball",
    "Doc 4: Italian Pasta",
    "Doc 5: Space Exploration",
    "Doc 6: Cybersecurity",
    "Doc 7: Music Theory",
    "Doc 8: Gardening",
    "Doc 9: Economics",
    "Doc 10: Public Health"
]

# 4. TfidfVectorizer fit aur transform karein
vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorizer.fit_transform(DOCUMENTS)

# 5. Sparse matrix ko Pandas DataFrame mein convert karna
df_tfidf = pd.DataFrame(
    tfidf_matrix.toarray(),
    index=doc_titles,
    columns=vectorizer.get_feature_names_out()
)

print("=== 1. Sample of the Matrix (First 3 Documents × First 10 Terms) ===")
# Displaying first 3 rows and first 10 columns
display(df_tfidf.iloc[:, :10])
print("\n" + "="*60 + "\n")

print("=== 2. Top-5 TF-IDF Terms for Each Document (Summary) ===")
feature_names = np.array(vectorizer.get_feature_names_out())

for i, title in enumerate(doc_titles):
    # Har document ke scores nikalna
    row_scores = tfidf_matrix.toarray()[i]

    # Sab se baray 5 scores ke indices nikalna (descending order mein)
    top_n_indices = row_scores.argsort()[::-1][:5]

    # Terms aur unke scores print karna
    top_terms = [(feature_names[idx], round(row_scores[idx], 4)) for idx in top_n_indices]

    print(f"\n{title}:")
    for term, score in top_terms:
        print(f"   - {term}: {score}")
```

```python
import os
import sys
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# 1. Agar repo clone nahi hui toh clone karein
if not os.path.exists('unstructuredata_tfidf_repo'):
    !git clone https://github.com/mateenyaqoob/unstructuredata_tfidf_repo.git

# 2. Path set kar ke documents import karein
sys.path.append('unstructuredata_tfidf_repo')
from data.documents import DOCUMENTS

# 3. Document titles define karein
doc_titles = [
    "Doc 1: AI Healthcare",
    "Doc 2: Climate Change",
    "Doc 3: Basketball",
    "Doc 4: Italian Pasta",
    "Doc 5: Space Exploration",
    "Doc 6: Cybersecurity",
    "Doc 7: Music Theory",
    "Doc 8: Gardening",
    "Doc 9: Economics",
    "Doc 10: Public Health"
]

# 4. TfidfVectorizer fit aur transform karein (Documents par)
vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorizer.fit_transform(DOCUMENTS)

# 5. Query Similarity Function define karna
def rank_documents_by_query(query, docs, vectorizer, doc_tfidf_matrix, titles):
    # Query ko usi vectorizer ke zariye transform karein (Fit nahi karna, sirf transform!)
    query_vector = vectorizer.transform([query])

    # Cosine similarity calculate karein query aur saare documents ke darmiyan
    similarity_scores = cosine_similarity(query_vector, doc_tfidf_matrix).flatten()

    # Documents ko similarity scores ke lihaz se descending order (highest to lowest) mein rank karein
    ranked_indices = similarity_scores.argsort()[::-1]

    print(f"\n==================================================")
    print(f"QUERY: '{query}'")
    print(f"==================================================")

    for rank, idx in enumerate(ranked_indices, start=1):
        score = similarity_scores[idx]
        print(f"{rank}. {titles[idx]} (Score: {score:.4f})")

# 6. Predefined queries par function run karna
queries = [
    "renewable energy and climate policy",
    "password security and phishing",
    "healthy sleep exercise and diet"
]

for q in queries:
    rank_documents_by_query(q, DOCUMENTS, vectorizer, tfidf_matrix, doc_titles)
```
