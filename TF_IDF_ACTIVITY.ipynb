2. Build the TF-IDF Matrix

Import TfidfVectorizer from scikit-learn.

Fit it on the documents using stop_words="english" to remove common English words.

Print the shape of the resulting matrix (expected: (10, n_unique_terms)) and the vocabulary size.

```python
# Initialize TfidfVectorizer with English stop words
vectorizer = TfidfVectorizer(stop_words='english')

# Fit and transform the documents
X = vectorizer.fit_transform(DOCUMENTS)

# Print matrix shape and vocabulary size
print("Matrix shape:", X.shape)
print("Vocabulary size:", len(vectorizer.get_feature_names_out()))
```

3. Inspect the Matrix as a DataFrame

Convert the sparse TF-IDF matrix into a pandas DataFrame with document titles as the index and terms as columns.

Print a sample of the matrix (e.g., the first 3 documents × 10 terms).

For each document, print the top-5 TF-IDF terms (the terms with the highest scores), which act as a summary of that document.


```python
import pandas as pd

# Convert the TF-IDF matrix to a pandas DataFrame
df_tfidf = pd.DataFrame(
    X.toarray(),
    columns=vectorizer.get_feature_names_out(),
    index=TITLES
)

# Print a sample of the matrix (first 3 documents and first 10 terms)
print("--- Sample of the TF-IDF Matrix (First 3 docs x 10 terms) ---")
display(df_tfidf.iloc[:3, :10].round(3))

# For each document, print the top-5 TF-IDF terms
print("\n--- Top 5 TF-IDF Terms Per Document ---")
for title in TITLES:
    print(f"\n{title}:")
    top5 = df_tfidf.loc[title].sort_values(ascending=False).head(5)
    for term, score in top5.items():
        print(f"  {term:<20s} {score:.3f}")
```

4. Query Similarity Challenge

Define a function that takes a query string, transforms it using the same vectorizer, and computes cosine similarity against all 10 documents.

Rank the documents by similarity score in descending order.

Run the function on three predefined queries and observe which documents rank highest:

"renewable energy and climate policy" → should rank Climate Change first.

"password security and phishing" → should rank Cybersecurity first.

"healthy sleep exercise and diet" → should rank Public Health first.

```python
from sklearn.metrics.pairwise import cosine_similarity

def rank_documents(query: str):
    """Rank and print documents based on cosine similarity to the query."""
    # Transform the query using the fitted vectorizer
    query_vec = vectorizer.transform([query])

    # Calculate cosine similarity between the query vector and the document matrix X
    scores = cosine_similarity(query_vec, X).flatten()

    # Pair each title with its score and sort in descending order
    rankings = sorted(zip(TITLES, scores), key=lambda x: x[1], reverse=True)

    print(f"\n=== Query: {query!r} ===")
    for title, score in rankings:
        print(f"  {title:<25s} {score:.3f}")

# Run the function on the three predefined queries
queries = [
    "renewable energy and climate policy",
    "password security and phishing",
    "healthy sleep exercise and diet"
]

for q in queries:
    rank_documents(q)
```

6. Verify Their Work

Run the provided unit tests with pytest -q. The tests check that:

10 documents are loaded and each is "long text" (>30 words).

The TF-IDF matrix has 10 rows and a column count equal to the vocabulary size.

The query "renewable energy and climate policy" ranks Climate Change first.

```python
"""Basic unit tests for the TF-IDF activity utilities."""
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Inline definitions of the helpers that the test expects:
def load_documents():
    # Using our in-memory data variables
    return TITLES, DOCUMENTS

def build_tfidf(docs):
    vectorizer = TfidfVectorizer(stop_words='english')
    matrix = vectorizer.fit_transform(docs)
    return vectorizer, matrix

def rank_documents_for_query(query, vectorizer, matrix, titles):
    query_vec = vectorizer.transform([query])
    scores = cosine_similarity(query_vec, matrix).flatten()
    rankings = sorted(zip(titles, scores), key=lambda x: x[1], reverse=True)
    return rankings

# --- Unit Tests ---
def test_load_documents():
    titles, docs = load_documents()
    assert len(titles) == 10
    assert len(docs) == 10
    assert all(len(d.split()) > 30 for d in docs)  # "long text"
    print("test_load_documents: PASSED")

def test_tfidf_matrix_shape():
    _, docs = load_documents()
    vectorizer, matrix = build_tfidf(docs)
    assert matrix.shape[0] == 10
    assert matrix.shape[1] == len(vectorizer.get_feature_names_out())
    print("test_tfidf_matrix_shape: PASSED")

def test_query_ranking_matches_expected_topic():
    titles, docs = load_documents()
    vectorizer, matrix = build_tfidf(docs)
    ranking = rank_documents_for_query(
        "renewable energy and climate policy", vectorizer, matrix, titles
    )
    assert ranking[0][0] == "Climate Change"
    print("test_query_ranking_matches_expected_topic: PASSED")

# Run the tests directly
test_load_documents()
test_tfidf_matrix_shape()
test_query_ranking_matches_expected_topic()
```
