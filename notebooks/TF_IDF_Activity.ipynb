## Step 3  Counting stopwords BEFORE removal
Using the English stopword list from `sklearn.feature_extraction.text`. Count occurrences in the original text; inspect specific words using a simple query function.

```python
import re
from collections import Counter
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS, TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

def tokenize(text):
    return re.findall(r"(?u)\b\w\w+\b", text.lower())

original_tokens = [tokenize(doc) for doc in documents]
stopword_counts = Counter(word for words in original_tokens for word in words if word in ENGLISH_STOP_WORDS)
print("Distinct stopwords found:", len(stopword_counts))
print("Total stopword occurrences:", sum(stopword_counts.values()))
print("All stopword frequencies:")
for word, count in stopword_counts.most_common():
    print(f"{word:18} {count}")
```

```python
# Query any stopword frequency before removal
def stopword_frequency(word, counts=stopword_counts):
    word = word.lower().strip()
    return counts.get(word, 0)

for word in ["the", "and", "is", "to", "of"]:
    print(f"{word:8} → {stopword_frequency(word)}")
```

## Step 4  Removing stopwords and VERIFYING
Keeping documents as plain strings (unstructured text). Printing before/after counts and show a cleaned sample.

```python
cleaned_tokens = [[word for word in words if word not in ENGLISH_STOP_WORDS] for words in original_tokens]
cleaned_documents = [" ".join(words) for words in cleaned_tokens]
for title, before, after in zip(titles, original_tokens, cleaned_tokens):
    print(f"{title:24} before={len(before):3} after={len(after):3} removed={len(before)-len(after):3}")
print("\nCleaned example:", cleaned_documents[0][:300], "...")
remaining = Counter(word for words in cleaned_tokens for word in words if word in ENGLISH_STOP_WORDS)
print("Remaining stopwords:", sum(remaining.values()))
assert sum(remaining.values()) == 0
for word in ["the", "and", "is", "to", "of"]:
    print(f"After removal, {word}: {remaining.get(word, 0)}")
```

## Step 5  TF-IDF matrix
Fitting on the original documents with `stop_words="english"`. The manually cleaned documents above are for demonstrating and verifying removal.

```python
vectorizer = TfidfVectorizer(stop_words="english")
X = vectorizer.fit_transform(documents)
terms = vectorizer.get_feature_names_out()
print("TF-IDF matrix shape:", X.shape)
print("Vocabulary size:", len(terms))
print("First 3 documents × first 10 terms (without DataFrame):")
print("Terms:", list(terms[:10]))
for i in range(3):
    print(titles[i], [round(float(v), 3) for v in X[i, :10].toarray()[0]])
```

```python
# Top 5 terms per document, directly from sparse matrix
for i, title in enumerate(titles):
    row = X.getrow(i).toarray().ravel()
    top_indices = row.argsort()[::-1][:5]
    print(title + ": " + ", ".join(f"{terms[j]} ({row[j]:.3f})" for j in top_indices))
```

## Task 4  Query Similarity Challenge
Using the **same fitted vectorizer** to transform the query, computing cosine similarity against the 10 documents, then rank all scores in descending order.

```python
def rank_documents_for_query(query, vectorizer, matrix, titles):
    query_vector = vectorizer.transform([query])
    scores = cosine_similarity(query_vector, matrix).ravel()
    return sorted(zip(titles, scores), key=lambda item: item[1], reverse=True)

queries = [
    ("renewable energy and climate policy", "Climate Change"),
    ("password security and phishing", "Cybersecurity"),
    ("healthy sleep exercise and diet", "Public Health"),
]

for query, expected in queries:
    ranking = rank_documents_for_query(query, vectorizer, X, titles)
    print(f"\nQUERY: {query}")
    for position, (title, score) in enumerate(ranking, start=1):
        print(f"{position:2}. {title:24} similarity={score:.4f}")
    print("Expected first:", expected, "| Actual first:", ranking[0][0])
    assert ranking[0][0] == expected
print("\nAll 3 query ranking checks PASSED.")
```
