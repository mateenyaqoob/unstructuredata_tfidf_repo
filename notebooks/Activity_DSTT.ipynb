```python
import nltk
import numpy as np

from src.load_data import load_documents
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

titles, documents = load_documents()

print("Number of titles:", len(titles))
print("Number of documents:", len(documents))
```

```python
for title, document in zip(titles, documents):
    words = nltk.tokenize.wordpunct_tokenize(document)
    print(title, "-", len(words), "words")
```

```python
vectorizer = TfidfVectorizer(stop_words="english")

tfidf_matrix = vectorizer.fit_transform(documents)

terms = vectorizer.get_feature_names_out()

print("TF-IDF Matrix Shape:", tfidf_matrix.shape)
print("Vocabulary Size:", len(terms))
```

```python
matrix = tfidf_matrix.toarray()

print("First 10 terms:", list(terms[:10]))
print("\nFirst 3 documents × First 10 terms:")

print(matrix[:3, :10])
```

```python
for i, title in enumerate(titles):
    scores = matrix[i]

    top_indices = np.argsort(scores)[::-1][:5]

    print("\nDocument:", title)
    print("Top 5 terms:")

    for j in top_indices:
        print(terms[j], ":", round(scores[j], 4))
```

```python
def search_documents(query):
    query_vector = vectorizer.transform([query])

    scores = cosine_similarity(
        query_vector, tfidf_matrix
    )[0]

    ranked_indices = np.argsort(scores)[::-1]

    print("\nQuery:", query)

    for i in ranked_indices:
        print(titles[i], ":", round(scores[i], 4))

    return ranked_indices
```

```python
search_documents("renewable energy and climate policy")
```

```python
search_documents("password security and phishing")
```

```python
search_documents("healthy sleep exercise and diet")
```
