```python
from src.load_data import load_documents

# Load document titles and text
titles, documents = load_documents()

# Check total documents
print("Total Documents:", len(documents))

# Print word count of each document
for title, doc in zip(titles, documents):
    print(f"{title}: {len(doc.split())} words")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(stop_words="english")

tfidf_matrix = vectorizer.fit_transform(documents)

print("TF-IDF Matrix Shape:", tfidf_matrix.shape)
print("Vocabulary Size:", len(vectorizer.get_feature_names_out()))
```

```python
# Display first 3 documents and first 10 TF-IDF values
print(tfidf_matrix[:3, :10].toarray())
```

```python
# Get vocabulary terms
feature_names = vectorizer.get_feature_names_out()

for i, title in enumerate(titles):
    scores = tfidf_matrix[i].toarray().flatten()

    top_indices = scores.argsort()[-5:][::-1]

    print(f"\nDocument: {title}")

    for index in top_indices:
        print(f"{feature_names[index]}: {scores[index]:.4f}")
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity_matrix = cosine_similarity(tfidf_matrix)

print("Cosine Similarity Matrix:")
print(similarity_matrix.round(3))
```

```python
import numpy as np

# Ignore self-similarity
similarity_copy = similarity_matrix.copy()
np.fill_diagonal(similarity_copy, 0)

# Find highest similarity
i, j = np.unravel_index(
    np.argmax(similarity_copy),
    similarity_copy.shape
)

print("Most Similar Documents:")
print(titles[i], "and", titles[j])
print("Similarity Score:", round(similarity_copy[i, j], 4))
```

```python
# Ignore diagonal values
similarity_copy = similarity_matrix.copy()
np.fill_diagonal(similarity_copy, np.inf)

# Find minimum similarity
i, j = np.unravel_index(
    np.argmin(similarity_copy),
    similarity_copy.shape
)

print("Least Similar Documents:")
print(titles[i], "and", titles[j])
print("Similarity Score:", round(similarity_copy[i, j], 4))
```

```python
import numpy as np

for i, title in enumerate(titles):
    scores = similarity_matrix[i].copy()
    scores[i] = -1  # Exclude the document itself

    best_match = np.argmax(scores)

    print(f"{title} -> {titles[best_match]}")
    print(f"Similarity: {scores[best_match]:.4f}\n")
```

```python
query = "artificial intelligence in medical diagnosis"

# Convert query into TF-IDF vector
query_vector = vectorizer.transform([query])

# Calculate similarity with all documents
query_scores = cosine_similarity(query_vector, tfidf_matrix).flatten()

# Rank documents by similarity
ranked_indices = np.argsort(query_scores)[::-1]

print("Search Query:", query)
print("\nDocument Rankings:")

for i in ranked_indices:
    print(f"{titles[i]}: {query_scores[i]:.4f}")
```

```python
queries = [
    "climate change and carbon emissions",
    "pasta cooking and Italian sauces",
    "computer security and cyber attacks"
]

for query in queries:
    query_vector = vectorizer.transform([query])
    scores = cosine_similarity(query_vector, tfidf_matrix).flatten()

    best_index = np.argmax(scores)

    print("Query:", query)
    print("Best Matching Document:", titles[best_index])
    print("Similarity Score:", round(scores[best_index], 4))
    print()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

def search_documents(query):
    query_vector = vectorizer.transform([query])
    scores = cosine_similarity(query_vector, tfidf_matrix).flatten()

    ranked_indices = np.argsort(scores)[::-1]

    print("\nQuery:", query)
    for i in ranked_indices:
        print(f"{titles[i]}: {scores[i]:.4f}")

queries = [
    "renewable energy and climate policy",
    "password security and phishing",
    "healthy sleep exercise and diet"
]

for query in queries:
    search_documents(query)
```
