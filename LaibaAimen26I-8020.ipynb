```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Create the TF-IDF vectorizer and remove English stop words
vectorizer = TfidfVectorizer(stop_words="english")

# Learn the vocabulary and calculate TF-IDF scores
tfidf_matrix = vectorizer.fit_transform(documents)

# Get the vocabulary
terms = vectorizer.get_feature_names_out()

print("TF-IDF matrix shape:", tfidf_matrix.shape)
print("Vocabulary size:", len(terms))
```

```python
import pandas as pd

# Convert the TF-IDF matrix into a table
df = pd.DataFrame(
    tfidf_matrix.toarray(),
    index=titles,
    columns=terms
)

# Display the first 3 documents and first 10 terms
display(df.iloc[:3, :10].round(3))
```

```python
for title in df.index:
    # Sort the document's word scores from highest to lowest
    top_terms = df.loc[title].sort_values(ascending=False).head(5)

    print(f"\n{title}")
    print(top_terms)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def rank_documents_for_query(query):
    # Transform the query using the SAME fitted vectorizer
    query_vector = vectorizer.transform([query])

    # Compare the query with all documents
    similarities = cosine_similarity(query_vector, tfidf_matrix).flatten()

    # Sort documents by similarity, highest first
    ranked_indices = similarities.argsort()[::-1]

    for i in ranked_indices:
        print(f"{titles[i]:25s} {similarities[i]:.3f}")


queries = [
    "renewable energy and climate policy",
    "password security and phishing",
    "healthy sleep exercise and diet"
]

for query in queries:
    print(f"\nQuery: {query}")
    rank_documents_for_query(query)
```

```python
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS
from collections import Counter
import re

total_stop_word_counts = Counter()

for title, document in zip(TITLES, DOCUMENTS):
    # Extract words and convert to lowercase
    words = re.findall(r"\b[a-zA-Z]+\b", document.lower())

    # Get stop words and count their frequency
    removed_words = [word for word in words if word in ENGLISH_STOP_WORDS]
    stop_word_counts = Counter(removed_words)

    # Remove stop words
    cleaned_words = [word for word in words if word not in ENGLISH_STOP_WORDS]
    cleaned_document = " ".join(cleaned_words)

    # Add this document's counts to the overall counts
    total_stop_word_counts.update(stop_word_counts)

    print(f"\nTitle: {title}")
    print("Stop words removed:", removed_words)
    print("Stop word frequencies:", dict(stop_word_counts))
    print("Cleaned document:", cleaned_document)

print("\n--- Overall Stop Word Frequency Across All Documents ---")
for word, frequency in total_stop_word_counts.most_common():
    print(f"{word}: {frequency}")
```
